```python
import os
import shutil

# ----------------------------------------------------
# [수정 필요] 작업할 상위 경로를 지정하세요.
# ----------------------------------------------------
base_path = "/SSD4/psleon/YOLOv11/data/Validation_total"
# ----------------------------------------------------


def merge_and_move_folders(source_parent, dest_parent):
    """
    source_parent 안의 모든 하위 폴더를 dest_parent로 이동시키고 병합합니다.
    """
    # 1. 원본 상위 폴더가 존재하는지 확인
    if not os.path.exists(source_parent):
        print(f"🟡 경고: 원본 폴더를 찾을 수 없습니다: {source_parent}")
        return False

    print(f"▶ '{os.path.basename(source_parent)}' -> '{os.path.basename(dest_parent)}' 작업을 시작합니다.")
    
    # 2. 원본 상위 폴더 내의 모든 항목(폴더) 목록을 가져옴
    try:
        subfolders = [f for f in os.listdir(source_parent) if os.path.isdir(os.path.join(source_parent, f))]
    except FileNotFoundError:
        print(f"🟡 원본 폴더 '{source_parent}'가 이미 없거나 접근할 수 없습니다.")
        return False

    if not subfolders:
        print("  -> 처리할 하위 폴더가 없습니다.")

    # 3. 각 하위 폴더에 대해 이동 및 병합 작업 수행
    for folder_name in subfolders:
        source_subfolder = os.path.join(source_parent, folder_name)
        dest_subfolder = os.path.join(dest_parent, folder_name)
        
        try:
            print(f"  - '{folder_name}' 폴더 이동 및 병합 중...")
            # 대상 폴더로 내용을 복사하며 덮어쓰기
            shutil.copytree(source_subfolder, dest_subfolder, dirs_exist_ok=True)
            # 원본 하위 폴더 삭제
            shutil.rmtree(source_subfolder)
        except Exception as e:
            print(f"  ❌ 오류 발생 ('{folder_name}'): {e}")
            
    # 4. 작업이 완료된 원본 상위 폴더(_copy) 삭제
    try:
        os.rmdir(source_parent)
        print(f"✅ 원본 상위 폴더 '{source_parent}'를 성공적으로 삭제했습니다.")
    except OSError:
        print(f"🟡 경고: '{source_parent}' 폴더가 비어있지 않아 삭제하지 못했습니다.")
        
    return True


# --- 메인 실행 코드 ---
if __name__ == "__main__":
    # 처리할 작업 목록 정의
    tasks = [
        {"source": "images_copy", "dest": "images"},
        {"source": "labels_copy", "dest": "labels"}
    ]

    for task in tasks:
        source_dir = os.path.join(base_path, task["source"])
        dest_dir = os.path.join(base_path, task["dest"])
        
        merge_and_move_folders(source_dir, dest_dir)
        print("-" * 30)

    print("\n🎉 모든 작업이 완료되었습니다.")
```
